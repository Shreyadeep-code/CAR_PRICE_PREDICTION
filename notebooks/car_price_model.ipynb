```python
df=df[df['year'].str.isnumeric()]
df['year'].unique()
```

```python
df['year']=df['year'].astype('int')
df.info()
```

```python
df['Price'].unique()
```

```python
df=df[df['Price']!='Ask For Price']
```

```python
df['Price']=df['Price'].str.replace(',','').astype('int')
df.info()
```

```python
df['kms_driven']=df['kms_driven'].str.split(' ').str.get(0).str.replace(',','')
df['kms_driven'].isna().sum()
```

```python
df=df[df['kms_driven'].str.isnumeric()]
```

```python

df.info()
```

```python
df['kms_driven']=df['kms_driven'].astype('int')
df.info()
```

```python
df=df[~df['fuel_type'].isna()]
df.info()
```

```python
df['name']=df['name'].str.split(' ') .str.slice(0,3).str.join(' ')
df
```

```python
df=df.reset_index(drop=True)
df
```

```python
df=df[df['Price']<6e6]
df
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(sparse_output=False)
ohe.fit(df[['name','company','fuel_type']])
```

```python
ohe.categories_
```

```python
X=df.drop('Price',axis=1)
y=df['Price']
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.svm import SVR
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
model1=LinearRegression()
model2=RandomForestRegressor()
model3=XGBRegressor()
model4=SVR()
```

```python
trf1=ColumnTransformer([
    ('one_hot',OneHotEncoder(categories=ohe.categories_),['name','company','fuel_type'])
],remainder='passthrough')
pipe1=make_pipeline(trf1,model1)
pipe2=make_pipeline(trf1,model2)
pipe3=make_pipeline(trf1,model3)
pipe4=make_pipeline(trf1,model4)
```

```python
pipe1.fit(x_train,y_train)
# pipe2.fit(x_train,y_train)
# pipe3.fit(x_train,y_train)
# pipe4.fit(x_train,y_train)
```

```python
pipe2.fit(x_train,y_train)
```

```python
 pipe3.fit(x_train,y_train)
```

```python
pipe4.fit(x_train,y_train)
```

```python
from sklearn.metrics import r2_score
```

```python
print('LinearRegression : ',r2_score(y_test,pipe1.predict(x_test)))
print('RandomForestRegressor : ',r2_score(y_test,pipe2.predict(x_test)))
print('XGBRegressor : ',r2_score(y_test,pipe3.predict(x_test)))
print('SVR : ',r2_score(y_test,pipe4.predict(x_test)))
```

```python
scores=[]
for i in range(1000):
    x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=i)
    pipe=make_pipeline(trf1,RandomForestRegressor())
    pipe.fit(x_train,y_train)
    scores.append(r2_score(y_test,pipe.predict(x_test)))
```
